**프롬프트의 3가지 요소**

프롬프트를 통해 모델에게 무엇을 해야 할지 가르치는 것을 인컨텍스트 학습(in-context learning)이라고 함. 컨텍스트 길이와 비용은 trade-off

작업 설명 (Instruction)
- 모델이 무엇을 해야하는지 설명.

예시 (One/Few-shot Examples)
- 기대하는 입출력이 어떤 모습인지 보여줌. 작업 설명만으로 부족할 때 큰 역할함.

실제 작업 (Task)
- 모델이 실제로 처리해야 할 입력.

**프롬프트 작성하기**

1. 명확하고 명시적인 지시 작성하기 (페르소나 포함)
2. 충분한 컨텍스트 제공하기
- 컨텍스트가 없으면 모델은 내부 지식으로 채움! 환각의 원인!
- RAG 파이프라인 등 모델이 검색하게 할 수도 있음.
3. 모델에게 생각할 시간 주기 (Step by Step)
- 지연 시간과 비용 (trade-off)
